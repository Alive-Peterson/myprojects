##### Overall Profitability

```python
# Overall profitability
total_revenue = df['Net_Revenue_USD'].sum()
total_cost = df['Total_Cost_USD'].sum()
total_profit = df['Profit_USD'].sum()

loss_orders = (df['Profit_USD'] < 0).sum()
profitable_orders = (df['Profit_USD'] > 0).sum()

print(f"Total Net Revenue: ${total_revenue:,.2f}")
print(f"Total Cost: ${total_cost:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Profitable Orders: {profitable_orders:,}")
print(f"Loss-Making Orders: {loss_orders:,}")
print(f"Loss-Making Order %: {loss_orders / len(df) * 100:.2f}%")
```

#### 1. Sales performance over time

```python
monthly_performance = (df.groupby('Year_Month').agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')).reset_index())

monthly_performance.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_performance['Year_Month'],
    monthly_performance['Revenue'],
    label='Revenue'
)

plt.plot(
    monthly_performance['Year_Month'],
    monthly_performance['Profit'],
    label='Profit'
)

plt.title('Monthly Revenue and Profit Trend')
plt.xlabel('Month')
plt.ylabel('USD')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

#### 2. Profitability Over Time

##### Monthly Profit

```python
monthly_profit = (
    df.groupby('Year_Month')['Profit_USD']
      .sum()
      .reset_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_profit['Year_Month'],
    monthly_profit['Profit_USD'],
    marker='o'
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Monthly Profit Trend')
plt.xlabel('Month')
plt.ylabel('Profit (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### Monthly Margin

```python
monthly_margin = (
    df.groupby('Year_Month')
      .agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
)

monthly_margin['Profit_Margin_Pct'] = (
    monthly_margin['Profit'] /
    monthly_margin['Revenue'] * 100
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_margin['Year_Month'],
    monthly_margin['Profit_Margin_Pct'],
    marker='o'
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Monthly Profit Margin Trend')
plt.xlabel('Month')
plt.ylabel('Profit Margin (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 3. Product/Category Performance

##### Revenue by Product Category

```python
category_performance = (
    df.groupby('Product_Category')
      .agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
)

category_performance
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    category_performance['Product_Category'],
    category_performance['Revenue']
)

plt.title('Net Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Net Revenue (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### Profit by Product Category

```python
category_performance_sorted = category_performance.sort_values(
    'Profit',
    ascending=False
)

plt.figure(figsize=(10, 5))

plt.bar(
    category_performance_sorted['Product_Category'],
    category_performance_sorted['Profit']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Profit by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Profit (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 5. Promotion Performance

```python
promotion_performance = (
    df.groupby('Promotion_Type')
      .agg(
          Orders=('Order_ID', 'count'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
      .sort_values('Profit', ascending=False)
)

promotion_performance
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    promotion_performance['Promotion_Type'],
    promotion_performance['Profit']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Profit by Promotion Type')
plt.xlabel('Promotion Type')
plt.ylabel('Profit (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 6. Customer / Business Segment Performance

##### B2B vs B2C profitability

```python
customer_performance = (
    df.groupby('Customer_Type')
      .agg(
          Orders=('Order_ID', 'count'),
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
)

customer_performance['Profit_Margin_Pct'] = (
    customer_performance['Profit'] /
    customer_performance['Revenue'] * 100
)

customer_performance
```

##### Profit by Customer Type

```python
plt.figure(figsize=(8, 5))

plt.bar(
    customer_performance['Customer_Type'],
    customer_performance['Profit']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Profit by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Profit (USD)')
plt.tight_layout()
plt.show()
```

#### 6. Loss making orders

```python
loss_summary = pd.DataFrame({
    'Metric': [
        'Total Orders',
        'Loss-Making Orders',
        'Loss-Making Order %',
        'Total Loss'
    ],
    'Value': [
        len(df),
        (df['Profit_USD'] < 0).sum(),
        (df['Profit_USD'] < 0).mean() * 100,
        df.loc[df['Profit_USD'] < 0, 'Profit_USD'].sum()
    ]
})

loss_summary
```

##### Loss Making Orders by Sales Channel

```python
channel_loss = (
    df.groupby('Sales_Channel')
      .agg(
          Orders=('Order_ID', 'count'),
          Loss_Orders=('Profit_USD', lambda x: (x < 0).sum())
      )
      .reset_index()
)

channel_loss['Loss_Order_Pct'] = (
    channel_loss['Loss_Orders'] /
    channel_loss['Orders'] * 100
)

channel_loss
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    channel_loss['Sales_Channel'],
    channel_loss['Loss_Order_Pct']
)

plt.title('Loss-Making Orders by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Loss-Making Orders (%)')
plt.tight_layout()
plt.show()
```

##### Total loss by channel

```python
channel_total_loss = (
    df[df['Profit_USD'] < 0]
    .groupby('Sales_Channel')['Profit_USD']
    .sum()
    .reset_index()
)

channel_total_loss
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    channel_total_loss['Sales_Channel'],
    channel_total_loss['Profit_USD']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Total Loss from Loss-Making Orders by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Total Loss (USD)')
plt.tight_layout()
plt.show()
```

#### Why online has loss orders?

```python
online_df = df[df['Sales_Channel'] == 'Online'].copy()
```

```python
online_df['Order_Status'] = np.where(
    online_df['Profit_USD'] < 0,
    'Loss-Making',
    'Profitable'
)
```

```python
online_comparison = (
    online_df.groupby('Order_Status')
    .agg(
        Orders=('Order_ID', 'count'),
        Avg_Revenue=('Net_Revenue_USD', 'mean'),
        Avg_COGS=('COGS_USD', 'mean'),
        Avg_Logistics=('Logistics_Cost_USD', 'mean'),
        Avg_Marketing=('Marketing_Spend_USD', 'mean'),
        Avg_Discount=('Discount_Amount_USD', 'mean'),
        Avg_Profit=('Profit_USD', 'mean'),
        Avg_Margin=('Profit_Margin_Pct', 'mean')
    )
    .round(2)
)

online_comparison
```

##### Marketing Spend vs Profit

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    online_df['Marketing_Spend_USD'],
    online_df['Profit_USD'],
    alpha=0.5
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Marketing Spend vs. Profit — Online Orders')
plt.xlabel('Marketing Spend (USD)')
plt.ylabel('Profit (USD)')
plt.tight_layout()
plt.show()
```

### Financial Performance and Profitability Analysis

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('datasets/bss_sales_data.csv')
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.dropna(how='all').copy()
```

```python
df.shape
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.duplicated().sum()
```

```python
df[['sku', 'salesdate']].duplicated().sum()
```

##### Checking if 596 sku records are duplicates or separate records

```python
dup = df[df[['sku', 'salesdate']].duplicated(keep=False)].sort_values(['sku', 'salesdate'])
dup.head(20)
```

```python
dup.groupby(['sku', 'salesdate']).size().value_counts()
```

```python
dup.groupby(['sku', 'salesdate']).agg(
    rows=('sku', 'size'),
    total_units=('unitsordered', 'sum'),
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum')
).sort_values('rows', ascending=False).head(20)
```

```python
df.dtypes
```

```python
df[['salesdate']].head()
```

```python
df[['price', 'unitsordered', 'sales', 'cogs', 'fba','reffee', 'adspend', 'profit']].describe()
```

##### Convert Data Types

###### For salesdate

```python
df['salesdate'] = pd.to_datetime(df['salesdate'])
```

```python
df['salesdate'].dtype
```

```python
print("Start:", df['salesdate'].min())
print("End:", df['salesdate'].max())
```

###### For unitsordered

```python
#checking whether it contains decimals
df[df['unitsordered'] % 1 != 0]['unitsordered'].count()
```

```python
df['unitsordered'] = df['unitsordered'].astype(int)
```

```python
df['unitsordered'].dtype
```

```python
calculated_profit = (
    df['sales']
    - df['cogs']
    - df['fba']
    - df['reffee']
    - df['adspend']
)

(df['profit'] - calculated_profit).abs().describe()
```

```python
(df['profit'] != calculated_profit).sum()
```

##### Checking Records in Sales and other numerical columns

###### Checking Zero sales records

```python
df[df['price'] == 0][['sku', 'salesdate', 'price', 'unitsordered', 'sales','cogs', 'fba', 'reffee', 'adspend', 'profit']].head(20)
```

```python
df[df['sales'] == 0][['sku', 'salesdate', 'price', 'unitsordered', 'sales','cogs', 'fba', 'reffee', 'adspend', 'profit']].head(20)
```

###### Checking Negative Values

```python
df[['price', 'unitsordered', 'sales', 'cogs', 'fba', 'reffee', 'adspend']].lt(0).sum()
```

###### Checking Pricing Boundaries

```python
(df['min_price'] > df['max_price']).sum()
```

```python
((df['price'] < df['min_price']) | (df['price'] > df['max_price'])).sum()
```

###### Checking if the records are 0

```python
df[df['min_price'] > df['max_price']][
    ['sku', 'salesdate', 'price', 'min_price', 'max_price',
     'unitsordered', 'sales', 'profit']
].head(20)
```

```python
boundary_error = df[df['min_price'] > df['max_price']]
(boundary_error['sales'] == 0).sum()
```

```python
below_min = df['price'] < df['min_price']
above_max = df['price'] > df['max_price']

print("Below minimum:", below_min.sum())
print("Above maximum:", above_max.sum())
```

```python
print("Zero price below minimum:", ((df['price'] == 0) & below_min).sum())
```

```python
boundary_error.groupby('sku').size().sort_values(ascending=False).head(20)
```

```python
boundary_error.groupby('sku').agg(
    rows=('sku', 'size'),
    min_price=('min_price', 'mean'),
    max_price=('max_price', 'mean'),
    avg_price=('price', 'mean'),
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum')
).sort_values('rows', ascending=False).head(20)
```

```python
nonzero_below_min = df[(df['price'] < df['min_price']) & (df['price'] > 0)]
nonzero_below_min.shape
```

```python
nonzero_below_min[
    ['sku', 'salesdate', 'price', 'min_price', 'max_price',
     'unitsordered', 'sales', 'profit']
].head(20)
```

```python
above_max_df = df[df['price'] > df['max_price']]

above_max_df[
    ['sku', 'salesdate', 'price', 'min_price', 'max_price',
     'unitsordered', 'sales', 'profit']
].head(20)
```

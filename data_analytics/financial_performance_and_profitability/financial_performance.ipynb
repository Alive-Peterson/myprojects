##### Checking Profit and Profit pct columns

```python
calculated_profit = (
    df['Net_Revenue_USD']
    - df['Marketing_Spend_USD']
    - df['COGS_USD']
    - df['Logistics_Cost_USD']
)

(df['Profit_USD'] - calculated_profit).abs().describe()
```

```python
calculated_margin = (
    df['Profit_USD'] / df['Net_Revenue_USD']
) * 100
(df['Profit_Margin_Pct'] - calculated_margin).abs().describe()
```

```python
df[['Net_Revenue_USD', 'Profit_USD', 'Profit_Margin_Pct']].head(10)
```

```python
calculated_margin = (
    df['Profit_USD'] / df['Net_Revenue_USD']
) * 100

pd.DataFrame({
    'Provided': df['Profit_Margin_Pct'],
    'Calculated': calculated_margin,
    'Difference': df['Profit_Margin_Pct'] - calculated_margin
}).head(10)
```

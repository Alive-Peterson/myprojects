##### Feature Engineering

###### Revenue per unit

```python
df['Revenue_per_Unit'] = df['Net_Revenue_USD'] / df['Units_Sold']
```

###### Total Cost

```python
df['Total_Cost_USD'] = (df['COGS_USD'] + df['Marketing_Spend_USD'] + df['Logistics_Cost_USD'])
```

###### Marketing ROI

```python
df['Marketing_ROI'] = (df['Profit_USD']/df['Marketing_Spend_USD'])
```

###### Discount Amount

```python
df['Discount_Amount_USD'] = (df['Gross_Sales_USD']*df['Discount_Pct']/100)
```

###### Converting Order_Date data type

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

###### Year-Month

```python
df['Year_Month'] = df['Order_Date'].dt.to_period('M').astype(str)
```

#### Inspecting and Analysing Loss 

```python
loss_df = df[df['Profit_USD'] < 0]

loss_df[['Units_Sold',
         'Unit_Price_USD',
         'Discount_Pct',
         'Gross_Sales_USD',
         'Marketing_Spend_USD',
         'COGS_USD',
         'Logistics_Cost_USD',
         'Net_Revenue_USD',
         'Profit_USD',
         'Profit_Margin_Pct']].describe()
```

```python
loss_df[['Product_Category',
         'Promotion_Type',
         'Sales_Channel',
         'Customer_Type']].describe(include='object')
```

```python
loss_df['Promotion_Type'].value_counts()
```

```python
loss_df['Sales_Channel'].value_counts()
```

```python
category_loss = df.groupby('Product_Category').agg(
    total_orders=('Order_ID', 'count'),
    loss_orders=('Profit_USD', lambda x: (x < 0).sum()),
    total_profit=('Profit_USD', 'sum')
).reset_index()

category_loss['loss_rate_pct'] = (
    category_loss['loss_orders'] /
    category_loss['total_orders'] * 100
)

category_loss.sort_values('loss_rate_pct', ascending=False)
```

```python
channel_loss = df.groupby('Sales_Channel').agg(
    total_orders=('Order_ID', 'count'),
    loss_orders=('Profit_USD', lambda x: (x < 0).sum()),
    total_profit=('Profit_USD', 'sum')
).reset_index()

channel_loss['loss_rate_pct'] = (
    channel_loss['loss_orders'] /
    channel_loss['total_orders'] * 100
)

channel_loss.sort_values('loss_rate_pct', ascending=False)
```

```python
customer_loss = df.groupby('Customer_Type').agg(
    total_orders=('Order_ID', 'count'),
    loss_orders=('Profit_USD', lambda x: (x < 0).sum()),
    total_profit=('Profit_USD', 'sum')
).reset_index()

customer_loss['loss_rate_pct'] = (
    customer_loss['loss_orders'] /
    customer_loss['total_orders'] * 100
)

customer_loss.sort_values('loss_rate_pct', ascending=False)
```

```python
promotion_loss = df.groupby('Promotion_Type').agg(
    total_orders=('Order_ID', 'count'),
    loss_orders=('Profit_USD', lambda x: (x < 0).sum()),
    total_profit=('Profit_USD', 'sum')
).reset_index()

promotion_loss['loss_rate_pct'] = (
    promotion_loss['loss_orders'] /
    promotion_loss['total_orders'] * 100
)

promotion_loss.sort_values('loss_rate_pct', ascending=False)
```

```python
product_loss = df.groupby('Product_Name').agg(
    total_orders=('Order_ID', 'count'),
    loss_orders=('Profit_USD', lambda x: (x < 0).sum()),
    total_profit=('Profit_USD', 'sum')
).reset_index()

product_loss['loss_rate_pct'] = (
    product_loss['loss_orders'] /
    product_loss['total_orders'] * 100
)

product_loss.sort_values('loss_rate_pct', ascending=False).head(10)
```

```python
channel_loss = df.groupby('Sales_Channel').agg(
    total_orders=('Order_ID', 'count'),
    loss_orders=('Profit_USD', lambda x: (x < 0).sum()),
    total_profit=('Profit_USD', 'sum')
).reset_index()

channel_loss['loss_rate_pct'] = (
    channel_loss['loss_orders'] / channel_loss['total_orders'] * 100
)

channel_loss.sort_values('loss_rate_pct', ascending=False)
```

##### Why online is losing  money?

```python
channel_financial = df.groupby('Sales_Channel').agg(
    orders=('Order_ID', 'count'),
    avg_units=('Units_Sold', 'mean'),
    avg_discount=('Discount_Pct', 'mean'),
    avg_marketing_spend=('Marketing_Spend_USD', 'mean'),
    avg_cogs=('COGS_USD', 'mean'),
    avg_logistics=('Logistics_Cost_USD', 'mean'),
    avg_net_revenue=('Net_Revenue_USD', 'mean'),
    avg_profit=('Profit_USD', 'mean'),
    avg_margin=('Profit_Margin_Pct', 'mean')
).reset_index()

channel_financial.sort_values('avg_profit')
```

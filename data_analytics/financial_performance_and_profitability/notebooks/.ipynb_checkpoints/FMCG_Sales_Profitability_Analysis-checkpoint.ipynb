##### Overall Profitability

```python
# Overall profitability
total_revenue = df['Net_Revenue_USD'].sum()
total_cost = df['Total_Cost_USD'].sum()
total_profit = df['Profit_USD'].sum()

loss_orders = (df['Profit_USD'] < 0).sum()
profitable_orders = (df['Profit_USD'] > 0).sum()

print(f"Total Net Revenue: ${total_revenue:,.2f}")
print(f"Total Cost: ${total_cost:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Profitable Orders: {profitable_orders:,}")
print(f"Loss-Making Orders: {loss_orders:,}")
print(f"Loss-Making Order %: {loss_orders / len(df) * 100:.2f}%")
```

#### 1. Sales performance over time

```python
monthly_performance = (df.groupby('Year_Month').agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')).reset_index())

monthly_performance.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_performance['Year_Month'],
    monthly_performance['Revenue'],
    label='Revenue'
)

plt.plot(
    monthly_performance['Year_Month'],
    monthly_performance['Profit'],
    label='Profit'
)

plt.title('Monthly Revenue and Profit Trend')
plt.xlabel('Month')
plt.ylabel('USD')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

#### 2. Profitability Over Time

##### Monthly Profit

```python
monthly_profit = (
    df.groupby('Year_Month')['Profit_USD']
      .sum()
      .reset_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_profit['Year_Month'],
    monthly_profit['Profit_USD'],
    marker='o'
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Monthly Profit Trend')
plt.xlabel('Month')
plt.ylabel('Profit (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### Monthly Margin

```python
monthly_margin = (
    df.groupby('Year_Month')
      .agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
)

monthly_margin['Profit_Margin_Pct'] = (
    monthly_margin['Profit'] /
    monthly_margin['Revenue'] * 100
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_margin['Year_Month'],
    monthly_margin['Profit_Margin_Pct'],
    marker='o'
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Monthly Profit Margin Trend')
plt.xlabel('Month')
plt.ylabel('Profit Margin (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 3. Product/Category Performance

##### Revenue by Product Category

```python
category_performance = (
    df.groupby('Product_Category')
      .agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
)

category_performance
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    category_performance['Product_Category'],
    category_performance['Revenue']
)

plt.title('Net Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Net Revenue (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### Profit by Product Category

```python
category_performance_sorted = category_performance.sort_values(
    'Profit',
    ascending=False
)

plt.figure(figsize=(10, 5))

plt.bar(
    category_performance_sorted['Product_Category'],
    category_performance_sorted['Profit']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Profit by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Profit (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 4. Sales Channel Performance

```python
channel_performance = (
    df.groupby('Sales_Channel')
      .agg(
          Revenue=('Net_Revenue_USD', 'sum'),
          Profit=('Profit_USD', 'sum')
      )
      .reset_index()
)

channel_performance['Profit_Margin_Pct'] = (
    channel_performance['Profit'] /
    channel_performance['Revenue'] * 100
)

channel_performance
```

##### Revenue by Sales Channel

```python
plt.figure(figsize=(8, 5))

plt.bar(
    channel_performance['Sales_Channel'],
    channel_performance['Revenue']
)

plt.title('Net Revenue by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Net Revenue (USD)')
plt.tight_layout()
plt.show()
```

##### Profit by Sales Channel

```python
plt.figure(figsize=(8, 5))

plt.bar(
    channel_performance['Sales_Channel'],
    channel_performance['Profit']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Profit by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Profit (USD)')
plt.tight_layout()
plt.show()
```

##### Profit Margin by Sales Channel

```python
plt.figure(figsize=(8, 5))

plt.bar(
    channel_performance['Sales_Channel'],
    channel_performance['Profit_Margin_Pct']
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.title('Profit Margin by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Profit Margin (%)')
plt.tight_layout()
plt.show()
```
